```python
# 1. 필요한 라이브러리 임포트
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 한글 폰트 설정 (Windows: Malgun Gothic, Mac: AppleGothic, Linux: NanumGothic)
plt.rcParams["font.family"] = "Malgun Gothic"
plt.rcParams["axes.unicode_minus"] = False

sns.get_dataset_names()
tips = sns.load_dataset('tips')
tips.info()
```

```python
fig = plt.figure(figsize=(14,6))
ax1 = fig.add_subplot(2,2,1)
ax2 = fig.add_subplot(2,2,2)
ax3 = fig.add_subplot(2,2,3)
ax4 = fig.add_subplot(2,2,4)

sns.stripplot(x='day', y='tip', hue='smoker', data=tips, alpha=0.7, ax=ax1)
sns.swarmplot(x='day', y='tip', hue='sex', data=tips, alpha=0.7, palette='Set2', size=4, ax=ax2)
sns.countplot(x='time', data=tips, alpha=0.7, ax=ax3)
sns.countplot(x='time', hue='day', data=tips, alpha=0.7, palette='Set1', ax=ax4)

ax1.set_title('날짜, 성별에 따른 Strip plot')
ax2.set_title('날짜, 성별에 따른 Swarm plot')
ax3.set_title('날짜, 성별에 따른 count plot - bar')
ax4.set_title('날짜, 성별에 따른 count plot-bar-day')
plt.show()
```

```python
fig = plt.figure(figsize=(12,5))
ax1 = fig.add_subplot(1,2,1)
ax2 = fig.add_subplot(1,2,2)

sns.regplot(x='total_bill', y='tip', data=tips, color='orangered', 
            scatter_kws={'s': 30, 'alpha': 0.4}, line_kws={'linestyle': '--'}, ax=ax1)
sns.regplot(x='total_bill', y='tip', data=tips, color='orangered', 
            scatter_kws={'s': 50, 'alpha': 0.6}, line_kws={'linestyle': '--'}, ax=ax2, fit_reg=False)
fig.suptitle("선형회기 그래프 그리기")
ax1.set_title("fit_reg 그리기, 기본")
ax2.set_title("fit_reg false, 설정 필요")
plt.show()
```

```python
fig = plt.figure(figsize=(12, 4))
ax1 = fig.add_subplot(1,2,1)

sns.histplot(tips['tip'], bins=20, kde=True, color='skyblue', stat='count', ax=ax1) # type: ignore
ax1.set_title("Histogram with KDE for Tips")
plt.show()
```

```python
sns.jointplot(x='size', y='tip', data=tips, kind='scatter')
plt.show()
```

```python
sns.pairplot(data=tips, hue='sex', diag_kind='hist', palette='husl')
plt.suptitle('Pairplot with Histograms by Gender', y=1.05)
plt.show()
```
